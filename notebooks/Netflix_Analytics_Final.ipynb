# Netflix Titles — End-to-End Analysis
This notebook loads, cleans, explores, and summarizes the Netflix Titles dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw/netflix_titles.csv')
df.shape
```

```python
df.info()
df.isna().sum().sort_values(ascending=False)
```

```python
clean=df.copy()
for c in ['director','cast','country','date_added','rating','duration']:
    clean[c]=clean[c].fillna('Unknown')
clean['date_added']=pd.to_datetime(clean['date_added'], errors='coerce')
clean['year_added']=clean['date_added'].dt.year
clean['listed_in_primary']=clean['listed_in'].str.split(',').str[0].str.strip()
clean['country_primary']=clean['country'].str.split(',').str[0].str.strip()
clean.head()
```

```python
clean['type'].value_counts()
```

```python
clean['release_year'].value_counts().sort_index().tail(20).plot(figsize=(10,4), marker='o', title='Recent Release-Year Trend'); plt.show()
```

```python
print('Top genres:')
print(clean['listed_in_primary'].value_counts().head(10))
print('Top countries:')
print(clean['country_primary'].value_counts().head(10))
```

## Key Takeaways
- The catalog contains both Movies and TV Shows, with Movies forming the larger share.
- Content is concentrated across a set of recurring genres and major production markets.
- Release-year and date-added trends can be used to study catalog growth and recency.
- Missing metadata is handled explicitly rather than silently dropped.
